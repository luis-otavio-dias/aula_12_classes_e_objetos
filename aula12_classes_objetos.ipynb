```python
class Casa:
  rua = "Rua 24"
  bairro = "Centro"
  cep = "1234"

  def enderecoCompleto(self):
    return self.rua + ", " + self.bairro + ", " + self.cep

casa = Casa()

print(casa.enderecoCompleto())
print(casa.rua)
print(casa.bairro)
print(casa.cep)
```
